```python
# ============================================================
# CELL 3 — KNN MAJOR RECOMMENDER
# ============================================================

from sklearn.neighbors import NearestNeighbors


def prepare_major_matrix():

    major_names = list(major_vectors.keys())

    major_matrix = np.array([
        [
            major_vectors[major][intelligence]
            for intelligence in intelligence_order
        ]
        for major in major_names
    ], dtype=float)

    # Major profiles are percentages (0–100)
    # Convert them to the same 0–1 scale as the user vector.
    major_matrix_normalized = major_matrix / 100.0

    return major_names, major_matrix_normalized


def recommend_majors(user_vector, k=3):

    major_names, major_matrix = prepare_major_matrix()

    k = min(k, len(major_names))

    knn = NearestNeighbors(
        n_neighbors=k,
        metric="euclidean"
    )

    knn.fit(major_matrix)

    distances, indices = knn.kneighbors(
        user_vector.reshape(1, -1)
    )

    results = []

    for rank, (distance, index) in enumerate(
        zip(distances[0], indices[0]),
        start=1
    ):

        major = major_names[index]

        # Maximum possible Euclidean distance
        # between two 8-dimensional 0–1 vectors.
        max_distance = np.sqrt(8)

        similarity = (
            1 - distance / max_distance
        ) * 100

        similarity = max(0, similarity)

        results.append({
            "Rank": rank,
            "Major": major,
            "Distance": round(float(distance), 4),
            "Similarity": round(float(similarity), 2)
        })

    return pd.DataFrame(results)


print("✓ KNN recommender ready")
print(f"✓ Comparing against {len(major_vectors)} majors")
```
